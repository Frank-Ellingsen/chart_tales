The Financial Times Visual Vocabulary organizes data visualization into 9 core narrative categories: Deviation, Correlation, Change v Time, Ranking, Distribution, Part to whole, Magnitude, Spatial, and Flow.
Each category serves a specific purpose in conveying information effectively through visual means. Here's a brief overview of each category:
1. **Deviation**: This category focuses on highlighting differences or variations from a standard or expected value. It is often used to show how data points deviate from a norm, such as in performance metrics or financial results.
2. **Correlation**: This category is used to illustrate relationships between two or more variables. It helps in identifying patterns, trends, and potential causations, making it useful for analyzing data sets where variables may influence each other.
3. **Change v Time**: This category emphasizes how data changes over a period of time. It is commonly used in line charts or time series graphs to track trends, growth, or decline in various metrics.
4. **Ranking**: This category is designed to show the relative position of items within a dataset. It is often represented through bar charts or tables, allowing for easy comparison of different entities based on specific criteria.
5. **Distribution**: This category focuses on the spread or dispersion of data points within a dataset. It helps in understanding the frequency and variability of data, often visualized through histograms or box plots.
6. **Part to Whole**: This category illustrates how individual components contribute to a larger whole. It is commonly represented through pie charts or stacked bar charts, providing insights into proportions and percentages within a dataset.
7. **Magnitude**: This category emphasizes the size or scale of data points, allowing for comparisons of absolute values. It is often visualized through bar charts, area charts, or bubble charts to convey the relative importance or impact of different elements.
8. **Spatial**: This category focuses on the geographical or spatial distribution of data. It is often represented through maps or geospatial visualizations, helping to identify patterns, clusters, or trends based on location.
9. **Flow**: This category illustrates the movement or transfer of data between different entities or stages. It is commonly visualized through flow charts, Sankey diagrams, or network graphs, providing insights into processes, relationships, and transitions within a system.
10. **Other**: This category encompasses visualizations that do not fit neatly into the other categories but still serve a valuable purpose in conveying information. It may include unique or innovative visual representations that enhance understanding and engagement with the data.
